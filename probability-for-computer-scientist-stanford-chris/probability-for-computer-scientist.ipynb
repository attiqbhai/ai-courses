## Sample Space for rolling two dice

```python
# Sample Space
sample_space = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
print("Sample space (all outcomes):")
print(sample_space)
print("\nTotal number of outcomes:", len(sample_space))

print("=========================================================")

# Event spave when sum is 7
event_space_sum_7 = [(d1, d2) for (d1, d2) in sample_space if d1 + d2 == 7]
print("Event space when sum is 7:")
print(event_space_sum_7)
print("\nTotal number of outcomes when sum is 7 :", len(event_space_sum_7))

print("=========================================================")

# Probability of Event Space
probability_sum_7 = len(event_space_sum_7) / len(sample_space)
print("Probability of sum = 7:", probability_sum_7)
```

## Or Probability with Non-Mutually Exclusive Events

```python
# Sample space for one die
sample_space = list(range(1, 7))
print(f"Sample Space: {sample_space}")

# Define events
A = {2, 4, 6}          # even
B = {4, 5, 6}          # > 3

# OR with non-mutually exclusive events
event_space = A.union(B)
overlap = A.intersection(B)

# Probability
prob = len(event_space) / len(sample_space)

print("A:", A)
print("B:", B)
print("Overlap A ∩ B:", overlap)
print("Event space A ∪ B:", event_space)
print("Probability:", prob)
```

## Inclusion Exclusion for Three Sets

$∣ 𝐴 ∪ 𝐵 ∪ 𝐶 ∣ = ∣ 𝐴 ∣ + ∣ 𝐵 ∣ + ∣ 𝐶 ∣ − ∣ 𝐴 ∩ 𝐵 ∣ − ∣ 𝐴 ∩ 𝐶 ∣ − ∣ 𝐵 ∩ 𝐶 ∣ + ∣ 𝐴 ∩ 𝐵 ∩ 𝐶 ∣$

```python
def inclusion_exclusion_three(A, B, C):
    return (len(A) + len(B) + len(C)
            - len(A & B) - len(A & C) - len(B & C)
            + len(A & B & C))

# Example sets
A = {2, 4, 6}
B = {3, 6}
C = {5}

result = inclusion_exclusion_three(A, B, C)
print("Size of A ∪ B ∪ C:", result)
```

## Conditional Probability

$𝑃 ( 𝐴 ∣ 𝐵 ) = \frac {𝑃 ( 𝐴 ∩ 𝐵 )}{ 𝑃 ( 𝐵 )}$

```python
# Sample space for two dice
sample_space = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
print(f"Sample space: {sample_space}")

# Define events
A = [(d1, d2) for (d1, d2) in sample_space if d1 + d2 == 7]      # sum = 7
B = [(d1, d2) for (d1, d2) in sample_space if d1 == 3]           # first die = 3

print(f"Event space 'A' when sum = 7 : {A}")
print(f"Event space 'B' when d1 = 3 : {B}")

# Intersection
A_and_B = [outcome for outcome in sample_space if outcome in A and outcome in B]
print(f"A ∩ B : {A_and_B}")

# Probabilities
P_A = len(A) / len(sample_space)
print(f"P ( A ) : {P_A}")
P_B = len(B) / len(sample_space)
print(f"P ( B ) : {P_B}")
P_A_and_B = len(A_and_B) / len(sample_space)
print(f"P ( A ∩ B ) : {P_A_and_B}")

# Conditional probability
P_A_given_B = P_A_and_B / P_B

print("P(A | B) =", P_A_given_B)
```

## Independence

$𝑃 ( 𝐴 ∣ 𝐵 ) = 𝑃 ( 𝐴 )$

```python
# Sample space for two dice
sample_space = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
print(f"Sample space: {sample_space}")

# Define events
A = [(d1, d2) for (d1, d2) in sample_space if d1 == 3]      # first die is 3
B = [(d1, d2) for (d1, d2) in sample_space if d1 + d2 == 7] # sum is 7

print(f"Event space 'A' when sum = 7 : {A}")
print(f"Event space 'B' when d1 = 3 : {B}")

# Probabilities
P_A = len(A) / len(sample_space)
P_B = len(B) / len(sample_space)
P_A_and_B = len([x for x in sample_space if x in A and x in B]) / len(sample_space)

# Conditional probability
P_A_given_B = P_A_and_B / P_B

# Check independence
independent = abs(P_A_and_B - P_A * P_B) < 1e-9

print("P(A):", P_A)
print("P(B):", P_B)
print("P(A ∩ B):", P_A_and_B)
print("P(A | B) =", P_A_given_B)
print("Independent:", independent)
```

## Chain Rule

For two events: 
> $𝑃 ( 𝐴 ∩ 𝐵 ) = 𝑃 ( 𝐴 )   𝑃 ( 𝐵 ∣ 𝐴 )$

For three events: 
> $𝑃 ( 𝐴 ∩ 𝐵 ∩ 𝐶 ) = 𝑃 ( 𝐴 )   𝑃 ( 𝐵 ∣ 𝐴 )   𝑃 ( 𝐶 ∣ 𝐴 ∩ 𝐵 )$

For n events: 
> $𝑃 ( 𝑋 1 , 𝑋 2 , … , 𝑋 𝑛 ) = 𝑃 ( 𝑋 1 )   𝑃 ( 𝑋 2 ∣ 𝑋 1 )   𝑃 ( 𝑋 3 ∣ 𝑋 1 , 𝑋 2 )   ⋯$

```python
# Sample space for two dice
sample_space = [(d1, d2) for d1 in range(1, 7) for d2 in range(1, 7)]
print(f"Sample space: {sample_space}")

# Events
A = [(d1, d2) for (d1, d2) in sample_space if d1 == 3]      # first die = 3
B = [(d1, d2) for (d1, d2) in sample_space if d1 + d2 == 7] # sum = 7

print(f"Event space 'A' when d1 = 3 : {A}")
print(f"Event space 'B' when sum = 7 : {B}")

# Compute probabilities
P_A = len(A) / len(sample_space)
P_B_given_A = len([x for x in A if x in B]) / len(A)
P_A_and_B_chain = P_A * P_B_given_A

print("P(A):", P_A)
print("P(B | A):", P_B_given_A)
print("Chain rule result:", P_A_and_B_chain)
```
