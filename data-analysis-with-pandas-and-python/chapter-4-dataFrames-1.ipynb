# Series Object


*   A pandas Series is a one-dimensional labelled array.
*   A Series combines the best features of a list and a dictionary.
*   A Series maintains a single collection of ordered values (i.e. a single column of data).
*   We can assign each value an identifier, which does not have to be unique.





```python
ice_cream_list = ["Chocolate", "Vanilla", "Strawbery", "Rum Raisin"]
ice_cream_series = pd.Series(ice_cream_list)
ice_cream_series
```

```python
registrations = [True, False, True, True, False, True]
registrations_series = pd.Series(registrations)

registrations_series
```

```python
A = pd.Series([2,5,7,9,11], index=["a", "b","c","d","e"])
print(type(A))
print(A)
print(type(A.values))
print(A.values)
print(type(A.index))
print(A.index)
print(A["a"])
print(A["a":"c"])
```

# Standard Deviation

Standard deviation is a measure of how spread out values are in a dataset.
*   Low standard deviation → values are close to the average
*   High standard deviation → values are widely spread out

Think of Standard deviation as the “typical distance” of data points from the mean.

#### Formal definition of Variance
*    Variance measures how spread out data values are from their mean (average).
*    Variance tells you how much the numbers differ from the average.
*    Small variance → values are close to the mean
*    Large variance → values are spread out

#### Formal definition of Standard deviation
Standard deviation is the square root of the variance.

#### Why use it?
*   Understand variability in data
*   Compare consistency (e.g., test scores, sales, sensor readings)
*   Core to statistics, machine learning, and data analysis

#### Rule of thumb (normal distribution)
*   ~68% of values lie within ±1 std dev of the mean
*   ~95% within ±2 std dev
*   ~99.7% within ±3 std dev

```python
prices = pd.Series([2.99, 4.45, 1.36])
print("-------The Standard Deviation-------")
print(prices.std())
```

# The sort_values Method



*   The `sort_values` method sorts a **Series** `values` in order.
*   By default, pandas applies **an ascending sort** (smallest to largest).
*   Customize the **sort order** with the `ascending` parameter.



```python
google = pd.read_csv("/content/google_stock_price.csv", usecols=["Stock Price"]).squeeze("columns")
print(google.sort_values().head(3))
print(google.sort_values(ascending=False).head(3))
```

# Extract Series Value by Index Position

*   Use the `iloc` accessor to extract a **Series** `value` by its `index` position.
*   `iloc` is short for "**index location**".
*   Python's list **slicing syntaxes (slices, slices from start, slices to end, etc.)** are supported with **Series** objects.



```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
print(pokemon.iloc[3])
print("---------------------------------------")
print(pokemon.iloc[[3,5,7]])
print("---------------------------------------")
print(pokemon.iloc[3:6])
print("---------------------------------------")
print(pokemon.iloc[0:4])
print("---------------------------------------")
print(pokemon.iloc[:2])
print("---------------------------------------")
print(pokemon.iloc[-4:])
```

# Extract Series Value by Index Label

*   Use the `loc` accessor to extract a **Series** `value` by its `index` label.
*   Pass a list to extract multiple values by `index` label.
*   If one `index` **label/position** in the **list** does not exist, Pandas will raise an error.


```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
print("---------------------------------------")
print(pokemon.loc["Bulbasaur"])
print("---------------------------------------")
print(pokemon.loc[["Bulbasaur","Ivysaur"]])
print("---------------------------------------")
```

# The get Method on a Series

*   The `get` method extracts a `Series` value by `index` label. It is an **alternative option to square brackets**.
*   The `get` method's second argument sets the **fallback value** to return if the **label/position** does not exist.


```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
print("---------------------------------------")
print(pokemon.get("Bulbasaur"))
print("---------------------------------------")
print(pokemon.loc["Bulbasaur"])
print("---------------------------------------")
print(pokemon.get("Bulbasaur", "Bulbasaur1"))
print("---------------------------------------")
```

# Overwrite a Series Value

*   Use the `loc/iloc` accessor to target an `index` **label/position**, then use an equal sign to provide a new value.



```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
print("---------------------------------------")
print(pokemon.get("Bulbasaur"))
print("---------------------------------------")
pokemon["Bulbasaur"] = "Fire"
print(pokemon.get("Bulbasaur"))
print("---------------------------------------")
print(pokemon.iloc[0])
print("---------------------------------------")
pokemon.iloc[0] = "Grass"
print(pokemon.get("Bulbasaur"))
print("---------------------------------------")
```

# Broadcasting

*   **Broadcasting** describes the process of applying **an arithmetic operation to an array** (i.e., a Series).
*   We can combine mathematical operators with a **Series** to apply the **mathematical operation** to **every value**.
*   There are also methods to accomplish the same results (**add, sub, mul, div, etc.**)


```python
google = pd.read_csv("/content/google_stock_price.csv", usecols=["Stock Price"]).squeeze("columns")
print(google.head(2))
print("-----------------------------------------")
google1 = google.add(10)
print(google1.head(2))
print("-----------------------------------------")
google2 = google1 - 10
print(google2.head(2))
print("-----------------------------------------")
google3 = google2 * 10
print(google3.head(2))
```

# The value_counts Method

*   The `value_counts` method returns the number of times **each unique value** occurs in the **Series**.
*   The `normalize` **parameter** returns the relative **frequencies/percentages of the values** instead of the **counts**.



```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
print("---------------------------------------")
print(pokemon.head(4))
print("---------------------------------------")
print(pokemon.value_counts().head(4))
print("---------------------------------------")
print(pokemon.value_counts(normalize=True).head(4))
```

# The map Method

*   The `map` method **"maps"** or **connects each Series values** to **another value**.
*   We can **pass** the **method** a **dictionary** or a **Series**. **Both** types **connects** **keys** to **values**.
*   The `map` method uses our **argument** to **connect or bridge together** the **values**.


```python
pokemon = pd.read_csv("/content/pokemon.csv", index_col=["Pokemon"]).squeeze("columns")
attach_powers = {
    "Grass": 10,
    "Fire": 15,
    "Dark": 30
}
print(pokemon.map(attach_powers).head(6))
```
