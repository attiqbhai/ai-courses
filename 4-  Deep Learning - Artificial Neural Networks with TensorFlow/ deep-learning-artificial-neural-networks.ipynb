# Hadrien Jean's Plot Vector Function

```python
def plot_vectors(vectors, colors, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(6,6))
        
    # Set the origin of the vectors to (0, 0)
    origin = np.array([[0], [0]])
    
    for i, vector in enumerate(vectors):
        ax.quiver(*origin, vector[0], vector[1], color=colors[i], 
                  angles='xy', scale_units='xy', scale=1, zorder=2)
        ax.text(vector[0], vector[1], f' v{i+1}', color=colors[i], 
                fontsize=14, verticalalignment='bottom')

    ax.set_xlim(-1, 5)
    ax.set_ylim(-1, 5)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_aspect('equal')
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)
    plt.show()

# Example usage:
# plot_vectors([[3, 4], [1, 2]], ['#1190FF', '#FF9A13'])
```
