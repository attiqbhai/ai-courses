```python
import pandas as pd
import kagglehub as kh
```

```python
s1 = pd.Series([11, 20, 30, 13, 25, 29])

print(s1)

print("--------------------")

print(s1.index)

print("--------------------")

print(s1.dtype)

print("--------------------")

print(s1[1])
```

```python
s2 = pd.Series([11, 20, 30, 13, 25, 29], index=['a', 'b', 'c', 'd', 'e', 'f'])

print(s2)

print("--------------------")

print(s2.index)

print("--------------------")

print(s2.dtype)

print("--------------------")

print(s2['b'])

print("--------------------")

print(s2['b':'d'])
```
