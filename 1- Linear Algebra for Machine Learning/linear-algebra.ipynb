# Matrices in PyTorch

```python
X_pt = torch.tensor([[25, 2], [5, 26], [3, 7]])

print(X_pt)

print(X_pt.shape)

print(X_pt[1,:])

print(X_pt[:,0])

print(X_pt.T)

print(X_pt * 2)

print(X_pt * 2 + 3)

print(torch.add(torch.multiply(X_pt, 2), 3))

Y_pt = X_pt + 2

print(Y_pt + X_pt)

print(Y_pt * X_pt)

```

# Dot Product

```python
X_pt = torch.tensor([25, 2., 7])
Y_pt = X_pt + 2

torch.dot(X_pt, Y_pt)
```
