```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

# Hadrien Jean's Plot Vector Function

```python
def plot_vectors(vectors, colors, ax=None):
    if ax is None:
        fig, ax = plt.subplots(figsize=(6,6))
        
    # Set the origin of the vectors to (0, 0)
    origin = np.array([[0], [0]])
    
    for i, vector in enumerate(vectors):
        ax.quiver(*origin, vector[0], vector[1], color=colors[i], 
                  angles='xy', scale_units='xy', scale=1, zorder=2)
        ax.text(vector[0], vector[1], f' v{i+1}', color=colors[i], 
                fontsize=14, verticalalignment='bottom')

    ax.set_xlim(-1, 5)
    ax.set_ylim(-1, 5)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_aspect('equal')
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)
    plt.show()

# Example usage:
# plot_vectors([[3, 4], [1, 2]], ['#1190FF', '#FF9A13'])
```

# Matrices in PyTorch

```python
X_pt = torch.tensor([[25, 2], [5, 26], [3, 7]])

print(X_pt)

print(X_pt.shape)

print(X_pt[1,:])

print(X_pt[:,0])

print(X_pt.T)

print(X_pt * 2)

print(X_pt * 2 + 3)

print(torch.add(torch.multiply(X_pt, 2), 3))

Y_pt = X_pt + 2

print(Y_pt + X_pt)

print(Y_pt * X_pt)

```

# Dot Product

```python
X_pt = torch.tensor([25, 2., 7])
Y_pt = X_pt + 2

torch.dot(X_pt, Y_pt)
```

# Matrix By Vector Multiplication

```python
A_pt = torch.tensor([[3, 4], [5, 6], [7, 8]])

b_pt = torch.tensor([2, 5])

print(torch.matmul(A_pt, b_pt))
```

# Matrix By Matrix Multiplication (In Matrix Multiplication -> A . B != B . A)

```python
A_pt = torch.tensor([[3, 4], [5, 6], [7, 8]])

B_pt = torch.tensor([[3, 5], [3, 2]])

print(torch.matmul(A_pt, B_pt))
```

# Symmetric Matrix

## Frobenius Norm
