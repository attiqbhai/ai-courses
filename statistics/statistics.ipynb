```python
import numpy as np
import pandas as pd
import torch
```

## Calculate Skew using Pandas

```python
data = pd.Series([1, 2, 2, 1, 3, 4, 3, 2, 5, 1, 2, 7, 3, 4, 5, 6, 2, 1])
print("Skewness:", data.skew())
```

## Caclculate Mean, Median and Mode

```python
data = pd.Series([1, 2, 2, 1, 3, 4, 3, 2, 5, 1, 2, 7, 3, 4, 5, 6, 2, 1])

print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
```

## Caclculate Range

```python
data = [4, 7, 10, 12, 20]
s = pd.Series(data)

range_value = s.max() - s.min()
```

## Calculate Interquartile Range

```python
data = [4, 7, 10, 12, 20]

# Convert to Pandas Series
s = pd.Series(data)

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)

# Interquartile Range
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Population and Sample

```python
population = np.arange(1, 1001)  # population of 100 values
sample = np.random.choice(population, size=100, replace=False)

print("Population mean:", np.mean(population))
print("Sample mean:", np.mean(sample))
print("Population Variance:", np.var(population))
print("Sample Variance:", np.var(sample))
```
