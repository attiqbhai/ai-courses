## Calculate Interquartile Range

```python
data = [4, 7, 10, 12, 20]

# Convert to Pandas Series
s = pd.Series(data)

# Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)

# Interquartile Range
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Population and Sample Variance

```python
population = np.arange(1, 1001)  # population of 100 values
sample = np.random.choice(population, size=100, replace=False)

print("Population mean:", np.mean(population))
print("Sample mean:", np.mean(sample))
print("Population Variance:", np.var(population)) # divide by N
print("Sample Variance:", np.var(sample, ddof=1))  # divide by n - 1
```

## Population and Sample Standard Deviation

```python
population = np.arange(1, 1001)  # population of 100 values
sample = np.random.choice(population, size=100, replace=False)

# Population standard deviation
population_sd = np.std(population)

# Sample standard deviation
sample_sd = np.std(sample, ddof=1)

print("Population SD:", population_sd)
print("Sample SD:", sample_sd)
```

## Scaling and Shifting

```python
data = np.array([10, 20, 30, 40, 50])

# -----------------------------
# Scaling factor and shift amount
# -----------------------------
scale_factor = 3
shift_amount = 10

# -----------------------------
# Apply scaling and shifting
# -----------------------------
scaled_data = data * scale_factor
shifted_data = data + shift_amount

# -----------------------------
# Function to compute stats
# -----------------------------
def stats(arr):
    return {
        "Mean": np.mean(arr),
        "Variance": np.var(arr),
        "Std Dev": np.std(arr),
        "Range": np.max(arr) - np.min(arr),
        "IQR": np.percentile(arr, 75) - np.percentile(arr, 25)
    }

# -----------------------------
# Calculate statistics
# -----------------------------
original_stats = stats(data)
scaled_stats = stats(scaled_data)
shifted_stats = stats(shifted_data)

# -----------------------------
# Display results
# -----------------------------
print("Original Data Stats:", original_stats)
print("\nScaled Data Stats:", scaled_stats)
print("\nShifted Data Stats:", shifted_stats)
```

## Z - Score and Z‑Table Probability

```python
data = np.array([10, 12, 13, 12, 11, 50])

mean = np.mean(data)
print(f"Mean is {mean}")

std = np.std(data)
print(f"Standard Deviation is {std}")

z_scores = (data - mean) / std

print(f"Z Scores are {z_scores}") # This gives you the Z‑score for every value.

prob = norm.cdf(z_scores)

print("Probability of Z Scores are :", prob)
```

## Basic Bayes’ Theorem in Python

```python
def bayes_theorem(P_A, P_B_given_A, P_B):
    return (P_B_given_A * P_A) / P_B

# Example values
P_A = 0.01          # Prior probability of A
P_B_given_A = 0.99  # Likelihood
P_B = 0.0594        # Total probability of B

posterior = bayes_theorem(P_A, P_B_given_A, P_B)
print("P(A|B) =", posterior)
```

## Bayes' - Full Medical Test Example (with total probability)

```python
# Prior probability
P_A = 0.01  # 1% have the disease

# Likelihoods
P_B_given_A = 0.99   # Test is 99% accurate
P_B_given_notA = 0.05  # 5% false positive rate

# Complement
P_notA = 1 - P_A

# Total probability of testing positive
P_B = P_B_given_A * P_A + P_B_given_notA * P_notA

# Bayes theorem
P_A_given_B = (P_B_given_A * P_A) / P_B

print("Probability of having the disease given a positive test:", P_A_given_B)
```

## Reusable Bayes Class (for ML or repeated calculations)

```python
class BayesCalculator:
    def __init__(self, prior, likelihood, false_positive):
        self.prior = prior
        self.likelihood = likelihood
        self.false_positive = false_positive

    def compute(self):
        P_A = self.prior
        P_notA = 1 - P_A
        P_B = self.likelihood * P_A + self.false_positive * P_notA
        return (self.likelihood * P_A) / P_B

# Example usage
bayes = BayesCalculator(prior=0.01, likelihood=0.99, false_positive=0.05)
print("Posterior:", bayes.compute())
```
